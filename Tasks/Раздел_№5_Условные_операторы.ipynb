## **Раздел №5. Условные операторы `if`, `elif`, `else` и `match case` в Python**

### **Зачем нужны условные операторы?**

**Условные операторы** позволяют выполнять разные блоки кода в зависимости от определённых условий. Это позволяет программам принимать решения и реагировать на различные ситуации.

---

````python
num = int(input('Введите число: '))

if num > 0:
  print('Ваше число больше нуля!')
elif num == 0:
    print('Ваше число ноль!')
else:
    print('Ваше число меньше нуля!')

print('THE END!')
````

### **Оператор `if`**

Оператор **`if`** используется для выполнения блока кода, если заданное условие истинно (`True`).

#### Синтаксис оператора `if`

````python
if условие:
    # Блок кода, выполняемый если условие истинно
````

**Важно:** В Python области видимости и блоки кода обозначаются отступами. Код, который должен выполняться внутри условия, записывается с отступом в 4 пробела от уровня `if`. Отступы позволяют понять, где начинается код, который должен выполняться при выполнении условия, и где заканчивается.

#### Пример использования `if`

````python
age = 18
if age >= 18:
    print("Вы совершеннолетний.")
````

**Вывод:**

````python
Вы совершеннолетний.
````

**Объяснение:**

- Если значение переменной `age` больше или равно `18`, условие `age >= 18` истинно.
- Выполняется блок кода внутри `if`.

### **Использование булевых значений в условных операторах**

В **условных выражениях**, таких как `if`, `elif`, а также циклах `while`, широко используются логические значения.

#### Пример:

````python
password = input("Введите пароль: ")

if password == "secret":
    print("Доступ открыт.")
else:
    print("Неверный пароль.")
````

#### Сложные булевские выражения

В `if` можно использовать более сложные булевские выражения с логическими операторами:

````python
x = 2
if (x > 3 and x < 11) or x == 5:
    print(x)
    # тоже выполнится при выполнении условия после if, так как этот код тоже записан с отступом в 4 пробела
    print("nice")
````

#### Оператор `is` и сравнение с `==`

Оператор **`is`** проверяет, ссылаются ли две переменные на **один и тот же объект** в памяти, в то время как **`==`** проверяет **равенство значений**.

##### Пример:

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a == b)  # True, значения равны
print(a is b)  # True, ссылаются на один объект
print(a == c)  # True, значения равны
print(a is c)  # False, разные объекты
````

**Важно:** Для сравнения значений используйте **`==`**, для проверки идентичности объектов — **`is`**.

#### Особенности выражений

##### Истинные и ложные значения

В Python некоторые значения по умолчанию считаются **ложными**:

- **`False`**
- **`None`**
- **`0`, `0.0`**
- **Пустые последовательности:** `''`, `[]`, `()`, `{}`
- **Пустые объекты**

Все остальные значения считаются **истинными**.

##### Пример:

````python
items = []

if items:
    print("Список не пустой.")
else:
    print("Список пустой.")
````

### **Оператор `if...else`**

Оператор **`else`** используется вместе с `if` для выполнения блока кода, когда условие в `if` ложно (`False`).

#### Синтаксис оператора `if...else`

````python
if условие:
    # Блок кода, если условие истинно
else:
    # Блок кода, если условие ложно
````

#### Пример использования `if...else`

````python
age = 16
if age >= 18:
    print("Вы совершеннолетний.")
else:
    print("Вы несовершеннолетний.")

# этот код уже не "внутри" else, потому что записан без отступа.
# Поэтому он выполнится в любом случае после отработки if-else
print("Done")
````

**Вывод:**

````python
Вы несовершеннолетний.
Done
````

**Объяснение:**

- Условие `age >= 18` ложно, поэтому выполняется блок кода внутри `else`.

### **Оператор `if...elif...else`**

Когда необходимо проверить несколько условий, используется оператор **`elif`** (сокращение от "else if"). Он позволяет добавить дополнительные условия после исходного `if`.

##### Синтаксис оператора `if...elif...else`

````python
if условие1:
    # Блок кода, если условие1 истинно
elif условие2:
    # Блок кода, если условие2 истинно
elif условие3:
    # Блок кода, если условие3 истинно
else:
    # Блок кода, если все условия ложны
````

##### Пример использования `if...elif...else`

````python
score = 75
if score >= 90:
    print("Оценка: Отлично")
elif score >= 75:
    print("Оценка: Хорошо")
elif score >= 60:
    print("Оценка: Удовлетворительно")
else:
    print("Оценка: Неудовлетворительно")



````

### **Вложенные условные операторы**

Условные операторы могут быть **вложены** друг в друга для проверки дополнительных условий.

##### **Пример вложенных условий**

````python
number = 10
if number > 0:
    print("Число положительное.")
    if number % 2 == 0:
        print("Число четное.")
    else:
        print("Число нечетное.")
else:
    print("Число отрицательное или ноль.")
````

##### **Пример с двойной вложенностью**

````python
x = 3
y = 5
if x == 3:
    if y > 6:
        # отступ в 4 пробела от внутреннего if
        print("y greater than 6")
    else:
        print("y not greater than 6")

    # отступ в 4 пробела от внешнего if, поэтому код выполнится если x==3 и при любом значении y
    print("x is equal to 3")
````

**Объяснение:**

- Проверяется, является ли число положительным.
- Если да, то внутри этого блока проверяется, является ли число четным.

### **Оператор `match case` (новинка Python 3.10)**

Оператор **`match case`** был введен в Python 3.10 и позволяет реализовать **структуру сопоставления с образцом**, похожую на оператор `switch` в других языках. Он используется для множественного выбора на основе значения переменной.

#### Синтаксис оператора `match case`



```python
match выражение:
    case шаблон1:
        # Блок кода, если выражение соответствует шаблону1
    case шаблон2:
        # Блок кода, если выражение соответствует шаблону2
    case _:
        # Блок кода по умолчанию
```



**Примечание:** Символ `_` используется как шаблон, который соответствует любому значению (аналог `default` в других языках).

#### Пример использования `match case`

````python
command = "start"

match command:
    case "start":
        print("Запуск программы...")
    case "stop":
        print("Остановка программы.")
    case "pause":
        print("Пауза программы.")
    case _:
        print("Неизвестная команда.")
````

**Объяснение:**

- Значение переменной `command` сравнивается с каждым `case`.
- При совпадении выполняется соответствующий блок кода.

---

#### **Сопоставление сложных шаблонов с `match case`**

Оператор `match case` поддерживает более сложное сопоставление, включая проверки на типы и значения вложенных структур данных.

##### Пример с сопоставлением списков

````python
point = (0, 0)

match point:
    case (0, 0):
        print("Точка находится в начале координат.")
    case (x, 0):
        print(f"Точка на оси X в позиции {x}.")
    case (0, y):
        print(f"Точка на оси Y в позиции {y}.")
    case (x, y):
        print(f"Точка в позиции X={x}, Y={y}.")
````

##### Задание 2: Калькулятор с `if...elif...else`

**Задача:** Напишите программу-калькулятор, которая запрашивает у пользователя два числа и операцию (`+`, `-`, `*`, `/`). Используйте `if...elif...else` для выполнения соответствующей операции и вывода результата.

##### **Решение:**

````python
num1 = float(input("Введите первое число: "))
num2 = float(input("Введите второе число: "))
operation = input("Выберите операцию (+, -, *, /): ")

if operation == '+':
    result = num1 + num2
    print(f"Результат: {result}")
elif operation == '-':
    result = num1 - num2
    print(f"Результат: {result}")
elif operation == '*':
    result = num1 * num2
    print(f"Результат: {result}")
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
        print(f"Результат: {result}")
    else:
        print("Ошибка: деление на ноль!")
else:
    print("Неверная операция.")
````

##### Задание 3: Использование `match case` для команд

**Задача:** Напишите программу, которая принимает от пользователя команду (`"вперед"`, `"назад"`, `"влево"`, `"вправо"`) и выводит соответствующее сообщение. Используйте оператор `match case`.

**(требуется Python 3.10 и выше):**

##### **Решение:**

````python
command = input("Введите команду (вперед, назад, влево, вправо): ")

match command:
    case "вперед":
        print("Двигаемся вперед.")
    case "назад":
        print("Отступаем назад.")
    case "влево":
        print("Поворачиваем налево.")
    case "вправо":
        print("Поворачиваем направо.")
    case _:
        print("Неизвестная команда.")
````
